## XBT times

```python
from datetime import datetime, timedelta

# ----------------------------
# USER INPUT
# ----------------------------

# Current time (24-hour format HH:MM)
current_time_str = "9:10"

# Time to add
add_hours = 13
add_minutes = 2

# ----------------------------
# CALCULATION
# ----------------------------

current_time = datetime.strptime(current_time_str, "%H:%M")
new_time = current_time + timedelta(hours=add_hours, minutes=add_minutes)

# ----------------------------
# OUTPUT
# ----------------------------

print("Current time:", current_time.strftime("%H:%M"))
print(f"After {add_hours} hours and {add_minutes} minutes:")
print("New time:", new_time.strftime("%H:%M"))
```
